```python
import matplotlib.pyplot as plt
import numpy as np

theta = np.linspace(0, 8 * np.pi, 1000)
r = theta
x = r * np.cos(theta)
y = r * np.sin(theta)

plt.figure(figsize=(7, 7))
plt.plot(x, y, color='purple', linewidth=1.5)
plt.axis('off')
plt.title("Spiral Pattern")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def draw_branch(ax, x, y, angle, length, depth):
    if depth == 0:
        return
    x2 = x + length * np.cos(np.radians(angle))
    y2 = y + length * np.sin(np.radians(angle))
    ax.plot([x, x2], [y, y2], color='saddlebrown', linewidth=depth)
    draw_branch(ax, x2, y2, angle - 20, length * 0.75, depth - 1)
    draw_branch(ax, x2, y2, angle + 20, length * 0.75, depth - 1)

fig, ax = plt.subplots(figsize=(8, 8))
draw_branch(ax, 0, 0, 90, 1, 10)
ax.set_xlim(-2, 2)
ax.set_ylim(0, 3)
ax.axis('off')
plt.title("Fractal Tree")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

n_points = 300
angles = np.linspace(0, 2 * np.pi, n_points)
x = np.cos(angles)
y = np.sin(angles)

# random scatter jitter for a "forming" look
jitter_x = x + np.random.normal(0, 0.02, n_points)
jitter_y = y + np.random.normal(0, 0.02, n_points)

plt.figure(figsize=(6, 6))
plt.scatter(jitter_x, jitter_y, c=angles, cmap='plasma', s=20)
plt.axis('off')
plt.title("Particle Formation")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np
from IPython.display import HTML

# Setup
n_points = 200
np.random.seed(42)

# Random starting positions (scattered)
start_x = np.random.uniform(-3, 3, n_points)
start_y = np.random.uniform(-3, 3, n_points)

# Target positions (a circle)
angles = np.linspace(0, 2 * np.pi, n_points)
target_x = np.cos(angles)
target_y = np.sin(angles)

fig, ax = plt.subplots(figsize=(6, 6))
scat = ax.scatter(start_x, start_y, c=angles, cmap='plasma', s=30)
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-3.5, 3.5)
ax.axis('off')

n_frames = 60

def update(frame):
    t = frame / n_frames  # progress from 0 to 1
    # smooth easing
    t_eased = t * t * (3 - 2 * t)
    current_x = start_x + (target_x - start_x) * t_eased
    current_y = start_y + (target_y - start_y) * t_eased
    scat.set_offsets(np.column_stack([current_x, current_y]))
    return scat,

ani = animation.FuncAnimation(fig, update, frames=n_frames, interval=50, blit=True)
plt.close()  # prevents duplicate static image
HTML(ani.to_jshtml())
```
